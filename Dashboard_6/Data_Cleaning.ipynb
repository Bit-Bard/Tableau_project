```python
import numpy as np
import pandas as pd
```

```python
data=pd.read_csv("Ola.csv")
data.head()
```

```python
data.shape
```

```python
data.info()
```

```python
data.isnull().sum()
```

```python
col_to_drop=['mta+AF8-tax','num+AF8-passengers','toll+AF8-amount','rate+AF8-code','stored+AF8-flag','improvement+AF8-charge','payment+AF8-method','driver+AF8-tip']

data=data.drop(col_to_drop, axis=1)
```

```python
data=data.rename(columns={
    'vendor+AF8-id':'Vendor ID',
    'pickup+AF8-loc': 'Pick_up Loc',
    'drop+AF8-loc': 'Drop Loc',
    'distance':'Distance',
    'pickup+AF8-time':'Pick_up Time',
    'drop+AF8-time':'Drop Time',
    'extra+AF8-charges':'Extra Charges',
    'total+AF8-amount':'Total Amount'
    
})
```

```python
data['Extra Charges'].unique()
```

```python
data.head()
```

```python
data.isnull().sum()
```

```python
data.shape
```

```python
data=data.dropna()
```

```python
dup=data.duplicated()
dup.sum()
data=data.drop_duplicates()
```

```python
data.shape
```

```python
data.info()
```

```python
data['Vendor ID']=data['Vendor ID'].astype(int)
```

```python
data['Extra Charges'].unique()
```

```python
data=data[data['Extra Charges']!='+AC0-0.5']
data=data[data['Extra Charges']!='+AC0-4.5']
data=data[data['Extra Charges']!= '+AC0-1']
```

```python
data['Extra Charges']=data['Extra Charges'].astype(float)
```

```python
data['Total Amount'].unique()
```

```python
data=data[data['Total Amount']!='+AC0-5.3']
data=data[data['Total Amount']!='+AC0-4.3']
data=data[data['Total Amount']!='+AC0-4.8']
data=data[data['Total Amount']!='+AC0-5.8']
data=data[data['Total Amount']!='+AC0-6.3']
data=data[data['Total Amount']!='+AC0-3.3']
data=data[data['Total Amount']!='+AC0-52.8']
data=data[data['Total Amount']!='+AC0-6.8']
```

```python
data = data[~data['Total Amount'].astype(str).str.contains('+AC0-', na=False, regex=False)]
```

```python
data['Total Amount']=data['Total Amount'].astype(float)
```

```python
data = data.reset_index(drop=True)
```

```python
data.shape
```

```python
data.info()
```

```python
data['Pick_up Time'] = pd.to_datetime(data['Pick_up Time'], format='%m/%d/%Y %I:%M:%S %p')
data['Drop Time'] = pd.to_datetime(data['Drop Time'], format='%m/%d/%Y %I:%M:%S %p')
```

```python
print(data['Pick_up Time'].dtype)
```

```python
data.shape
```

**Feature Extraction**

```python
data['Trip Duration (min)']=((data['Drop Time']-data['Pick_up Time']).dt.total_seconds()/60).astype(int)
```

```python
data['Hour of Day']=data['Pick_up Time'].dt.hour
```

```python
data['Day of Week']=data['Pick_up Time'].dt.day_name()
```

```python
data.head()
```

```python
data.columns
```

```python
data.to_csv("Cleaned_Data.csv",index=False)
```

```python
#there are huge number of rows in it so it will create problems while making Dashboard so we are doing sampling to reduce it
```

```python
sample_size = 10000

df_sampled = data.groupby('Hour of Day', group_keys=False).apply(lambda x: x.sample(frac=sample_size/len(data), random_state=42))
```

```python
df_sampled.shape
```

```python
df_sampled.head()
```

```python
# Say your data has:

# 10 lakh rows
# Hour 8 has 50,000 rows (5%)
# Hour 9 has 1,00,000 rows (10%
# If you ask for a 100,000-row sample (sample_size=100000), this line ensures:
# Hour 8 contributes ~5% of 100,000 → ~5,000 rows
# Hour 9 contributes ~10% of 100,000 → ~10,000 rows
# …and so on for all hours.
```

```python
df_sampled.to_csv("Filtered_Data.csv", index=False)
```
